```python
# imports
import pandas as pd
```

```python
# main datasets and the other datasets to be merged

main_df = pd.read_csv('disney_movies.csv')
other_df = pd.read_csv('movies.csv')
```

```python

other_df = other_df.rename(columns={'title': 'movie_title'})
main_df['movie_title'] = main_df['movie_title'].astype(str).str.strip()
other_df['movie_title'] = other_df['movie_title'].astype(str).str.strip()


snow_white_check = other_df[other_df['movie_title'].str.contains('Snow White', case=False, na=False)]
print("Found in additional dataset:")
print(snow_white_check[['movie_title', 'genres']])


other_subset = other_df[['movie_title', 'genres']].drop_duplicates(subset=['movie_title'])

merged_df = pd.merge(main_df, other_subset, on='movie_title', how='left')
merged_df['genre'] = merged_df['genres'].where(merged_df['genres'].notna(), merged_df['genre'])
merged_df = merged_df.drop(columns=['genres'])

merged_df.to_csv('disney_movies.csv', index=False)

print("Data cleaning and merging completed successfully. The updated dataset has been saved to 'disney_movies.csv'.")
display(merged_df.head())

```
